```python
def sum_of_list(list):
    total = 0
    for i in range (len(list)):
        total = total + list[i]
    return total



list = [1, 2, 3, 4, 5]
result = sum_of_list(list)
print(f"the sum of 5 natural numbers are : {result}")
```

```python
def sum_of_list(list):
    n = len(list)
    total = (n*(n+1))/2
    return total



list = [1, 2, 3, 4, 5]
result = sum_of_list(list)
print(f"the sum of 5 natural numbers are : {result}")
```
